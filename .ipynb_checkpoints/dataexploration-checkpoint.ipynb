```python
import pandas as pd

raw_business_data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales_USD': [12500, 14800, 13250, 15900, 17100, 18500], 
    'Region': ['North', 'North', 'South', 'East', 'West', 'West']
}

df_sample = pd.DataFrame(raw_business_data)

print("Sales Data Preview:")
display(df_sample)

df_sample.to_csv("sales_records.csv", index=False)

print("✅ File Created: 'sales_records.csv' is saved and ready for audit!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read data from the CSV file into a virtual memory grid
df = pd.read_csv("sales_records.csv")

print("--- EXCEL-STYLE PREVIEW (FIRST 3 ROWS) ---")
print(df.head(3))

print("\n--- INSTANT STATISTICAL EXECUTIVE SUMMARY ---")
print(df.describe())

# 2. Configure a visual graph layout window (10 inches wide, 5 inches high)
plt.figure(figsize=(10, 5))

# 3. Choose chart type (bar), match columns to X/Y axes, and set a hex color code
plt.bar(df['Month'], df['Sales_USD'], color='#1b5e20')

# 4. Inject context labels directly onto the presentation canvas layout
plt.title("H1 Regional Gross Revenue Trends", fontsize=14, fontweight='bold')
plt.xlabel("Reporting Period (Months)", fontsize=11)
plt.ylabel("Total Inflow (USD $)", fontsize=11)

# 5. Render the chart directly on our screen inside the notebook
plt.show()
```
