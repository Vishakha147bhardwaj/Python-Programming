```python
import json
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://data.worldbank.org/'
response = requests.get(url)
print('Response is successful',response.status_code)
```

```python
soup = BeautifulSoup(response.content,"html.parser")
print('Printed the soup content, title', soup.title.text)
```

```python
world_data = soup.find_all('div',class_='card')
print('total number of card',len(world_data))
```

```python
world_bank = []

for bank in world_data:
    title = bank.find("h3")

    ratio = bank.find("h2")

    world_bank.append({
        "Title": title,
        "Ratio": ratio
    })

print("Total cards extracted:", len(world_bank))
    
```
