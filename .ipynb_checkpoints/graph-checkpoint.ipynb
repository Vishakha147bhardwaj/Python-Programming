```python
import matplotlib.pyplot as plt

# Sample data: E-commerce sales segments
categories = ['Electronics', 'Clothing', 'Home Decor', 'Books', 'Beauty']
sales_count = [450, 320, 180, 290, 110]

# Create figure
plt.figure(figsize=(8, 5))

# Plot bars with a clean color palette
plt.bar(categories, sales_count, color='#3498db', edgecolor='black')

# Enhance readability
plt.title('Units Sold by Product Category', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Units Sold', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)  # Grid lines behind bars
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate dummy timeline (12 months)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
revenue_2025 = [12000, 14500, 13000, 17000, 19500, 22000, 21000, 23000, 25000, 28000, 31000, 35000]

plt.figure(figsize=(10, 5))

# Plot line with markers at data points
plt.plot(months, revenue_2025, marker='o', color='#2ecc71', linewidth=2.5, label='Monthly Revenue ($)')

plt.title('2025 Revenue Growth', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Revenue ($)', fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate 500 random delivery times centered around 30 minutes
np.random.seed(42) 
delivery_times = np.random.normal(loc=30, scale=5, size=500)

plt.figure(figsize=(8, 5))

# 'bins' divides the data range into intervals
plt.hist(delivery_times, bins=25, color='#e74c3c', edgecolor='white', alpha=0.8)

plt.title('Food Delivery Time Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Delivery Time (minutes)', fontsize=12)
plt.ylabel('Frequency (Order Count)', fontsize=12)
plt.grid(axis='y', alpha=0.3)

plt.show()
```
