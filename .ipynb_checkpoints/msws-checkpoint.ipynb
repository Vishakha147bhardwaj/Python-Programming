```python
import json
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://data.worldbank.org/'
response = requests.get(url)
print('Response is successful',response.status_code)
```

```python
soup = BeautifulSoup(response.content,"html.parser")
print('Printed the soup content, title', soup.title.text)
```

```python
world_data = soup.find_all('div',class_='card')
print('total number of card',len(world_data))
```

```python
world_bank = []

for bank in world_data:
    title = bank.find("h3")

    ratio = bank.find("h2")

    world_bank.append({
        "Title": title,
        "Ratio": ratio
    })

print("Total cards extracted:", len(world_bank))
    
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate dummy timeline (12 months)
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
revenue_2025 = [12000, 14500, 13000, 17000, 19500, 22000, 21000, 23000, 25000, 28000, 31000, 35000]

plt.figure(figsize=(10, 5))

# Plot line with markers at data points
plt.plot(months, revenue_2025, marker='o', color='#2ecc71', linewidth=2.5, label='Monthly Revenue ($)')

plt.title('2025 Revenue Growth', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Revenue ($)', fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```
