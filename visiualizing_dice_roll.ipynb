```python
import random

# Run Python interpreter visualization setup
random.seed(42)
num_rolls = 1000
results = []

for _ in range(num_rolls):
    results.append(random.randint(1, 6) + random.randint(1, 6))

# Count occurrences to map data explicitly
unique_sums = list(range(2, 13))
frequencies = [results.count(s) for s in unique_sums]

print("Dice Sum Frequencies:", dict(zip(unique_sums, frequencies)))
```

```python

import random
import matplotlib.pyplot as plt

# Set up random number generation
random.seed(42)

num_rolls = 1000
results = []

# Simulate rolling two dice 1,000 times
for _ in range(num_rolls):
    results.append(random.randint(1, 6) + random.randint(1, 6))

# Count occurrences of each possible sum
unique_sums = list(range(2, 13))
frequencies = [results.count(s) for s in unique_sums]

# Print the frequency table
print("Dice Sum Frequencies:", dict(zip(unique_sums, frequencies)))

# Plot the histogram
plt.figure(figsize=(10, 5))

plt.hist(
    results,
    bins=[i + 0.5 for i in range(2, 14)],
    edgecolor="black"
)

plt.xticks(unique_sums)
plt.xlabel("Sum of Two Dice")
plt.ylabel("Frequency")
plt.title("Histogram of Two Dice Rolls (1,000 Simulations)")
plt.grid(axis="y", alpha=0.3)

plt.show()
```
