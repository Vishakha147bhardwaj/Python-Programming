```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate clean mathematical data points
x = np.linspace(0, 10, 100)
y_sine = np.sin(x)
y_cosine = np.cos(x)

# 2. Use the Object-Oriented approach to set up a figure container
fig, ax = plt.subplots(figsize=(8, 4))

# 3. Plot continuous lines with custom colors, widths, and style accents
ax.plot(x, y_sine, color='royalblue', linewidth=2.5, label='Sine Wave')
ax.plot(x, y_cosine, color='darkorange', linewidth=2.0, linestyle='--', label='Cosine Wave')

# 4. Overlay a scatter plot onto the same canvas to point out specific highlights
sample_x = np.array([1, 3, 5, 7, 9])
sample_y = np.sin(sample_x)
ax.scatter(sample_x, sample_y, color='crimson', s=50, zorder=3, label='Sample Checkpoints')

# 5. Add text context, clean up boundaries, and enable the legend
ax.set_title("Trigonometric Functions & Checkpoints", fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel("Time / Phase (Radians)", fontsize=11)
ax.set_ylabel("Amplitude", fontsize=11)
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower left', frameon=True)

# Display the final plot window clean of system logs
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create clean mock data for a sales dashboard
categories = ['Electronics', 'Clothing', 'Home Gear', 'Books', 'Toys']
sales_revenue = [54000, 42000, 21000, 12000, 31000]

# Generate random normal distribution data to simulate user ratings
user_ratings = np.random.normal(loc=4.1, scale=0.5, size=1000)

# Set up a multi-plot layout grid side-by-side (1 row, 2 distinct plot columns)
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))

# --- PLOT 1: Categorical Bar Chart ---
# Using a color map array to make distinct categories pop out visually
bar_colors = ['#1f77b4', '#aec7e8', '#ff7f0e', '#ffbb78', '#2ca02c']
bars = ax1.bar(categories, sales_revenue, color=bar_colors, edgecolor='black', alpha=0.85)

# Dynamically annotate the exact numerical value above each bar container
for bar in bars:
    height = bar.get_height()
    ax1.annotate(f"${height:,.0f}",
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 3),  # Offset 3 points vertically
                 textcoords="offset points",
                 ha='center', va='bottom', fontsize=9)

ax1.set_title("Revenue by Product Department", fontsize=12, fontweight='bold')
ax1.set_ylabel("Total Income (USD)")
ax1.set_xticklabels(categories, rotation=15) # Rotate text to avoid overlapping

# --- PLOT 2: Numerical Histogram ---
# Divides continuous numerical data into discrete bins to show volume distribution
counts, bins, patches = ax2.hist(user_ratings, bins=20, color='teal', edgecolor='white', alpha=0.75)
ax2.axvline(x=np.mean(user_ratings), color='red', linestyle='-.', linewidth=1.5, label=f"Average ({np.mean(user_ratings):.2f})")

ax2.set_title("Distribution of Customer Ratings", fontsize=12, fontweight='bold')
ax2.set_xlabel("Review Score (1.0 - 5.0)")
ax2.set_ylabel("Frequency Count")
ax2.legend()

plt.tight_layout() # Prevents charts, labels, and tickmarks from overlapping 
plt.show()
```
